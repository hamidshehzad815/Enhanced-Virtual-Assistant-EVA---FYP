# CLF1 — Input Type Classifier Test
Tests: `command`, `confirmation`, `general`, `help`, `multiple_command`, `out_of_scope`

```python
import torch
from transformers import DistilBertTokenizerFast, DistilBertForSequenceClassification

MODEL_PATH = r'D:\FYP_20March\Eva_AI_Server\models\clf1_model'

tokenizer = DistilBertTokenizerFast.from_pretrained(MODEL_PATH)
model     = DistilBertForSequenceClassification.from_pretrained(MODEL_PATH)
model.eval()

id2label = model.config.id2label
print('Model loaded.')
print('Labels:', id2label)
```

```python
# ── Add your test inputs here ──────────────────────────────────────
inputs = [
    "add physics exam on monday at 3pm",
    "yes",
    "no cancel it",
    "hi there",
    "what can you do",
    "order a pizza",
    "add math exam and also delete physics lab",
    "help me",
    "delete all done tasks and show pending ones",
    "okay go ahead",
]
```

```python
print(f"{'Input':<50} {'Predicted':<20} {'Confidence'}")
print('-' * 85)

for text in inputs:
    enc = tokenizer(
        text.lower(),
        return_tensors = 'pt',
        truncation     = True,
        max_length     = 64,
    )
    with torch.no_grad():
        logits = model(**enc).logits
    probs      = torch.softmax(logits, dim=-1).squeeze()
    pred_id    = probs.argmax().item()
    pred_label = id2label[pred_id]
    confidence = probs[pred_id].item()
    print(f"{text:<50} {pred_label:<20} {confidence:.4f}")
```

```python
# ── Detailed probabilities for all classes ─────────────────────────
text = inputs[0]   # change index to inspect a specific input

enc = tokenizer(text.lower(), return_tensors='pt', truncation=True, max_length=64)
with torch.no_grad():
    logits = model(**enc).logits
probs = torch.softmax(logits, dim=-1).squeeze()

print(f'Input: "{text}"\n')
for i, prob in enumerate(probs.tolist()):
    print(f'  {id2label[i]:<20}: {prob:.4f}')
```
